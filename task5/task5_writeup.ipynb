InfoNCE (Information Noise Contrastive Estimation) is a contrastive loss that teaches the model to understand which image and caption belong together. It does this by comparing the cosine similarity between every image-text pair in a batch. During training, the model learns to increase the similarity of the correct image-caption pair while decreasing the similarity of all the incorrect pairs. Unlike standard classification, where the model chooses one label from a fixed set of classes, InfoNCE does not predict class labels. Instead, it learns meaningful embeddings by comparing similarities, making it useful for matching related images and text.

The temperature controls how confident the softmax distribution is. A low temperature makes the softmax very sharp, so the model strongly favors the highest similarity. This encourages the correct image-text pair to dominate, but if the correct pair is not already the most similar, the loss becomes very large and training can become unstable. On the other hand, a very high temperature makes the softmax almost uniform, so the model can no longer clearly distinguish between correct and incorrect pairs. The loss approaches a constant value of log(N), the gradients become very small, and learning slows down. Instead of choosing a fixed value, CLIP learns the temperature during training so it can automatically find the right balance between being too confident and not confident enough.

L2 normalization ensures that all embeddings have unit length, so the dot product directly represents the cosine similarity between them. This means the model compares embeddings based only on their direction rather than their magnitude. If normalization is skipped, the model could increase the magnitude of the embeddings to obtain larger similarity scores instead of learning meaningful representations. As a result, the similarity scores become unreliable, training can become unstable, and the model may minimize the loss by simply producing larger vectors rather than actually learning better image-text alignment.

In contrastive learning, every other sample in the batch acts as a negative example. Therefore, increasing the batch size increases the number of negatives that the model must distinguish from the correct image-text pair. This makes the learning task more challenging and encourages the model to learn more discriminative and robust embeddings instead of relying on simple or accidental similarities. As a result, larger batch sizes often lead to better representation learning. However, the catch is that larger batches require significantly more memory and computation, making training slower and more expensive.
