 *   When we calculate attention, we take the dot product between the query (Q) and key (K) vectors. As the size of these vectors dk increases, the dot product values also tend to grow larger because we are summing more numbers together. If Q and K are treated like random vectors with mean 0 and variance 1, the variance of their dot product increases roughly in proportion to d
k
	​

  *   These large attention scores are then passed into the softmax function. The problem is that softmax reacts very strongly to large values. Even a small difference between two scores can make one probability become very close to 1 while all the others become almost 0. This makes the attention distribution extremely sharp and reduces the flow of useful gradients during training.


 * To avoid this issue, we divide the attention scores by
sqrt of dk. This scaling keeps the values in a smaller and more stable range before applying softmax. As a result, the softmax outputs remain smoother and the model can learn more effectively instead of becoming overly confident about a single position too early in training.

*   In causal self-attention, the model should only look at the current token and the tokens before it. It must never “peek” at future tokens while predicting the next word.

*   If we apply the mask after softmax by simply setting future attention values to zero, there is a problem. During the softmax calculation, the future positions were still included when computing the probabilities. This means some probability mass was already assigned to those future tokens. After zeroing them out, the remaining probabilities no longer add up to 1, making the attention distribution incorrect.

* Instead, the mask is applied before softmax by setting future-position scores to negative infinity. In softmax, e
−∞
 becomes essentially 0, so future tokens automatically receive zero attention. At the same time, the valid positions are still normalized properly, so the probabilities remain meaningful.

* Q, K, and V stand for Query, Key, and Value, and they are the main parts that make the attention mechanism work.

* An intuitive way to think about them is through Google search analogy. Say you search something on Google, that search text is the Query(Q) which represents what you are looking for.

* Now, Google has millions of webpages stored and each webpage contains many keywords, topics etc. They are similar to Keys(K). Google compares the Query to the keys to figure out what output to give out.

* It is done mathematically using the dot product Q.K^T which gives out attention scores and higher scores means higher relevance.
The model uses these scores to figure which tokens are important and retrieves information linked to them - this is the Value(V).

* From a linear algebra POV :
 the Query projection learns what the token wants to find,
the Key projection learns what information the token can offer,
and the Value projection learns what information should actually be shared with other tokens.

* The single-head attention model performs slightly better than the bigram model because it can look at all previous characters in the context window instead of just the immediately previous character.

* A bigram model only learns relationships between two consecutive characters. For example, after seeing “q”, it may learn that “u” is likely to come next, but it cannot remember anything beyond one step. Because of this, it captures only very short-range patterns and usually produces highly random or repetitive text.

* The single-head attention model, on the other hand, can attend to multiple previous characters at once. This allows it to capture slightly richer context and learn simple patterns that span across several characters. Even though the generated text is still mostly gibberish, it tends to have somewhat more consistent structure compared to the pure bigram model.

* The reason the improvement is only marginal is that the model is still extremely limited. The main bottlenecks are model capacity and depth. A single attention head with a small embedding size cannot learn complex language structure. The model is also shallow, meaning it does not have multiple layers to gradually build higher-level understanding.
