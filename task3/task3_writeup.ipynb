* The CNN baseline achieved a higher validation accuracy (74.99%) compared to the Vision Transformer (64.35%) on CIFAR-10. This result is expected because CNNs have strong built-in biases for image processing. Through convolutional filters, CNNs focus on local regions of an image and learn hierarchical features such as edges, textures, shapes, and eventually complete objects.
* In contrast, the Vision Transformer treats an image as a sequence of patches and uses self-attention to model relationships between all patches. However, unlike CNNs, a ViT does not have built-in assumptions about image structure and must learn these patterns directly from the data. As a result, ViTs typically require larger datasets and more training data to achieve their full potential.
* Therefore, on a relatively small dataset such as CIFAR-10, the CNN outperformed the ViT because its inductive biases provided a significant advantage, while the ViT's flexibility was not fully utilized due to the limited amount of training data.



8)
 * Patching is necessary in a Vision Transformer because feeding individual pixels directly into the transformer would create an extremely long sequence. For a 32×32 RGB image, there are 32 × 32 × 3 = 3072 pixel values. Since self-attention has a computational complexity of O(T²), where T is the sequence length, computing attention between all pixels would require millions of attention scores, making it computationally expensive and inefficient.
* Additionally, individual pixels contain very little semantic information on their own. A single pixel does not tell us much about the image, whereas a small patch contains local patterns such as edges, textures, and shapes.
* Therefore, instead of treating every pixel as a token, ViTs divide the image into patches and treat each patch as a token, similar to how words are treated as tokens in a language model. Self-attention is then computed between patches rather than individual pixels, making the model much more efficient while still allowing it to learn relationships across the entire image.

9)
* The CLS token acts as a dedicated summary token for the entire image. At the beginning of the forward pass, it does not contain any meaningful information. However, during each self-attention layer, the CLS token can attend to all image patches and gather information from them. Over multiple transformer blocks, it gradually accumulates information about the entire image and becomes a learned representation of the image as a whole.
* While averaging all patch tokens is also possible, averaging treats every patch equally and may dilute important information. The CLS token, on the other hand, can learn through attention to focus more on the most relevant patches and ignore less important ones. As a result, it often provides a better representation for classification tasks.

10)
* In Task 2, a causal mask was necessary because the transformer was being used for language modeling. When predicting the next token, the model should only be allowed to look at the current and previous tokens, not future ones. The causal mask enforces this by preventing a token from attending to positions that come later in the sequence.
* In Task 3, the transformer is used for image classification. An image is divided into patches, and all patches are available at the same time. Unlike text generation, there is no concept of "future" patches in an image. Therefore, every patch should be able to attend to every other patch in order to gather as much information as possible about the entire image.
* If the causal mask were accidentally kept in the Vision Transformer, each patch would only be able to attend to patches that appear earlier in the patch sequence. This would artificially restrict the flow of information across the image.
As a result, the model would fail to capture many important relationships between different regions of the image, leading to worse classification performance.

11)
* Position embeddings for image patches encode the spatial location of each patch within the original image. After an image is split into patches and converted into tokens, the transformer only sees a sequence of patch embeddings and has no inherent knowledge of where each patch came from.
* Without position embeddings, the model would not be able to distinguish between a patch from the top-left corner and a patch from the bottom-right corner if their visual content were similar. As a result, important spatial relationships within the image would be lost.
* Position embeddings for image patches play the same role as position embeddings in text: they provide information about the arrangement of the input elements so that the transformer can interpret them correctly.
