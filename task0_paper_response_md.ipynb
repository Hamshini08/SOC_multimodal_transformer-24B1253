1) The authors were trying to improve sequence models used for tasks like machine translation. Before the Transformer, most models were based on RNNs or LSTMs, which processed words one at a time and were slow to train. Some of these models also used attention mechanisms, but they still depended on recurrence. The Transformer replaced recurrence completely and relied only on self-attention, which allowed the model to process sequences in parallel, learn long-range relationships better, and train much faster.

2) Self-attention computes how relevant other words in a sequence are to a particular word. For this, every word is embedded with a vector [Q,K,V] where Q represents the query of the word. This query is compared with the queries of other words for computing the similarity scores (they are computed by Q.K^T). These similarity scores are scaled and passed through a softmax function to turn them into probabilites and attention weights. The weights determine how much importance each word should receive and they are used to compute a weighted combination of the value vectors. This gives a new representation for every word that contains information from other relevant words in the sentence. If the output of these value vector representations is high, that means this particualr value should be outputted with respect to the query.
