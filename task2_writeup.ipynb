24) Attention helps each token look at and gather relevant information from other tokens in the sequence. For example, a word can attend to other words that provide useful context for understanding its meaning. However, attention mainly focuses on exchanging information between different positions. The MLP then processes each token independently and combines the different features within that token's representation. Through its nonlinear transformations, the MLP can learn more complex patterns and relationships between features, allowing the model to build richer representations than attention alone can provide.

25) I used the pre-norm transformer architecture, where LayerNorm is applied before the attention and MLP sublayers. In the implementation, this is done through self.attn(self.ln1(x)) and self.ffwd(self.ln2(x)). LayerNorm normalizes the activations of each token so that they remain on a stable scale, which helps prevent training instability. Pre-norm is generally easier to train for deep networks because the residual connection provides a direct path for gradients to flow backward through the network. In contrast, post-norm applies LayerNorm after the residual addition, causing gradients to pass through more operations and making optimization more difficult as the network depth increases. As a result, pre-norm transformers typically exhibit more stable training and better gradient flow.

Without these shortcuts, gradients must pass through every attention and MLP layer sequentially, making optimization much more difficult and slowing convergence.
